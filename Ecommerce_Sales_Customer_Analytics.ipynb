```python
query = """
SELECT order_status, COUNT(*) as order_count
FROM orders
GROUP BY order_status
ORDER BY order_count DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT o.order_id, o.order_status, c.customer_city, c.customer_state
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT p.product_category_name,
       COUNT(oi.order_id) as items_sold,
       ROUND(SUM(oi.price), 2) as total_revenue
FROM order_items oi
JOIN products p ON oi.product_id = p.product_id
GROUP BY p.product_category_name
ORDER BY total_revenue DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Day 4: SQL - CTEs & Window Functions
Used CTEs and window functions to identify top customers by spending and rank states by revenue

```python
query = """
WITH customer_spending AS (
    SELECT o.customer_id, SUM(p.payment_value) as total_spent
    FROM orders o
    JOIN payments p ON o.order_id = p.order_id
    GROUP BY o.customer_id
)
SELECT customer_id, total_spent
FROM customer_spending
ORDER BY total_spent DESC
LIMIT 10;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT c.customer_state,
       ROUND(SUM(p.payment_value), 2) as state_revenue,
       RANK() OVER (ORDER BY SUM(p.payment_value) DESC) as revenue_rank
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN payments p ON o.order_id = p.order_id
GROUP BY c.customer_state
ORDER BY revenue_rank
LIMIT 10;
"""
pd.read_sql(query, conn)
```

```python
query = """
WITH monthly_revenue AS (
    SELECT strftime('%Y-%m', o.order_purchase_timestamp) as month,
           ROUND(SUM(p.payment_value), 2) as revenue
    FROM orders o
    JOIN payments p ON o.order_id = p.order_id
    GROUP BY month
)
SELECT month, revenue,
       ROUND(SUM(revenue) OVER (ORDER BY month), 2) as running_total
FROM monthly_revenue
ORDER BY month;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT strftime('%Y-%m', order_purchase_timestamp) as month, COUNT(*) as order_count
FROM orders
GROUP BY month
ORDER BY month;
"""
pd.read_sql(query, conn)
```

```python
query = """
WITH monthly_revenue AS (
    SELECT strftime('%Y-%m', o.order_purchase_timestamp) as month,
           ROUND(SUM(p.payment_value), 2) as revenue
    FROM orders o
    JOIN payments p ON o.order_id = p.order_id
    WHERE strftime('%Y-%m', o.order_purchase_timestamp) NOT IN ('2016-09', '2016-12', '2018-09', '2018-10')
    GROUP BY month
)
SELECT month, revenue,
       ROUND(SUM(revenue) OVER (ORDER BY month), 2) as running_total
FROM monthly_revenue
ORDER BY month;
"""
pd.read_sql(query, conn)
```

## Day 5: Python EDA & Statistics
Analyzed delivery time patterns, payment methods, and review scores; tested whether delivery time affects customer satisfaction

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
orders["delivery_time_days"]=(orders["order_delivered_customer_date"]-orders["order_purchase_timestamp"]).dt.days
orders["delivery_time_days"].describe()
```

```python
orders[orders['delivery_time_days'] > 100][['order_id', 'order_status', 'order_purchase_timestamp', 'order_delivered_customer_date', 'delivery_time_days']].sort_values('delivery_time_days', ascending=False).head(10)
```

```python
orders[orders['order_delivered_customer_date'].dt.date == pd.to_datetime('2017-09-19').date()].shape[0]
```

```python
orders_clean_delivery = orders[orders['order_delivered_customer_date'].dt.date != pd.to_datetime('2017-09-19').date()]

orders_clean_delivery['delivery_time_days'].describe()
```

```python
orders_clean_delivery['order_delivered_customer_date'].dt.date.value_counts().head(10)
```

```python
plt.figure(figsize=(10,6))
sns.histplot(orders_clean_delivery['delivery_time_days'], bins=50, kde=True)
plt.xlabel('Delivery Time (days)')
plt.title('Distribution of Delivery Times')
plt.show()
```

```python
delivery_review = orders_clean_delivery.merge(reviews, on='order_id')

delivery_review[['delivery_time_days', 'review_score']].corr()
```

```python
import numpy as np

n = len(delivery_review)
r = -0.340765

t_stat = r * np.sqrt((n - 2) / (1 - r**2))
print("t-statistic:", t_stat)
print("Sample size:", n)
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='review_score', y='delivery_time_days', data=delivery_review)
plt.xlabel('Review Score')
plt.ylabel('Delivery Time (days)')
plt.title('Delivery Time by Review Score')
plt.show()
```
